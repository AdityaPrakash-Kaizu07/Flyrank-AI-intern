## 2. The question: decision, action, cost of a wrong call

The Search Question: Which content pages should be prioritized for refresh, expansion, or pruning to maximize engagement and ROI?

**Unit of Analysis**: A single content page (URL).
The Output: A ranked queue featuring a suggested action, a priority score, and reason codes per page.

**The Decision & Action:**
Decision: Optimizes how the editorial team allocates limited hours across a large content catalog.

**Action**: Editors work down the ranked queue to execute the suggested updates.

**Cost of a Wrong Call:**
False Positive (Updating a low-impact page): Wasted editorial hours and opportunity cost.

**False Negative** (Ignoring a high-value, decaying page): Lost organic traffic and degrading search visibility.

**Why ML Helps**: Simple rules (e.g., "update pages over a year old") miss real-world context. ML combines multiple signals—like age, CTR drops, and demand—to prioritize updates based on actual opportunity rather than rigid heuristics.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load the starter dataset
df = pd.read_csv('/content/Flyrank-AI-intern/data/raw/content_refresh_anonymized.csv')

# 1. Total scale of the catalog
total_pages = df['content_id'].nunique()

# 2. Stale Content: Pages not updated in over a year (365 days)
stale_pages = len(df[df['days_since_last_update'] > 365])
stale_pct = (stale_pages / total_pages) * 100

# 3. Active Decay: Pages where traffic is currently trending downward
decaying_pages = len(df[df['trend_pct'] < 0])
decaying_pct = (decaying_pages / total_pages) * 100

print(f"Total pages: {total_pages:,}")
print(f"Stale pages (untouched > 1 yr): {stale_pages:,} ({stale_pct:.1f}%)")
print(f"Pages with declining traffic: {decaying_pages:,} ({decaying_pct:.1f}%)")
```
